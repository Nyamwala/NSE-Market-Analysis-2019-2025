#### Core KPI Calculations

**Calculating daily returns and annualized volatility for each stock**

```python
# Sort values by stock Code and Date to ensure correct chronological order
df_master = df_master.sort_values(by=['Code', 'Date'])

# Calculate daily percentage returns based on 'Day Price'
df_master['Daily_Return'] = df_master.groupby('Code')['Day Price'].pct_change()

# Quick check to verify daily returns are calculating
print(df_master[['Date', 'Code', 'Day Price', 'Daily_Return']])
```

**Annualized volatility(measures how much an asset's returns fluctuate over a one-year period, expressed as an annualized standard deviation to quantify investment risk).**

```python
# Calculate Annualized Volatility per stock name and sector
df_volatility = (
    df_master.groupby(['Name', 'Sector'])['Daily_Return']
    .std()
    .reset_index(name='Daily_Volatility')
)
df_volatility['Annualized_Volatility'] = df_volatility[
    'Daily_Volatility'
] * np.sqrt(252)

# Sort by highest volatility to see which companies are the riskiest
df_volatility = df_volatility.sort_values(
    by='Annualized_Volatility', ascending=False
)

print('--- Top 10 Most Volatile Companies (2019-2025) ---')
print(df_volatility.head(10))
```

### Compound Annual Growth Rate (CAGR) and Maximum Drawdown ($MaxDD$)

**These two KPIs tell you the absolute growth story and the worst-case shock experienced by each company between 2019 and 2025 (capturing major events like the COVID-19 crash and market reforms).**

```python
# 1. Calculate CAGR (Compound Annual Growth Rate) for each company
# First, get the chronological first and last available price for each company
df_sorted = df_master.sort_values(by=['Name', 'Date'])

# Drop rows where Day Price might be missing/NaN
df_clean_price = df_sorted.dropna(subset=['Day Price'])

first_prices = df_clean_price.groupby('Name').first().reset_index()
last_prices = df_clean_price.groupby('Name').last().reset_index()

# Merge first and last prices
df_cagr = pd.merge(
    first_prices[['Name', 'Sector', 'Date', 'Day Price']],
    last_prices[['Name', 'Date', 'Day Price']],
    on='Name',
    suffixes=('_start', '_end'),
)

# Calculate number of years (approximate based on date difference)
df_cagr['Years'] = (
    pd.to_datetime(df_cagr['Date_end'])
    - pd.to_datetime(df_cagr['Date_start'])
).dt.days / 365.25

# Avoid division by zero for stocks with very short history
df_cagr['Years'] = df_cagr['Years'].apply(lambda x: x if x > 0 else 1)

# CAGR formula: (Ending Price / Beginning Price)^(1 / Years) - 1
df_cagr['CAGR'] = (
    df_cagr['Day Price_end'] / df_cagr['Day Price_start']
) ** (1 / df_cagr['Years']) - 1

# Sort by highest CAGR
df_cagr = df_cagr.sort_values(by='CAGR', ascending=False)

print('--- Top 10 Compounding Growth Companies (CAGR) ---')
print(
    df_cagr[
        ['Name', 'Sector', 'Day Price_start', 'Day Price_end', 'Years', 'CAGR']
    ].head(10)
)
```

**Maximum Drawdown**
 - is a critical risk metric because it measures the largest peak-to-trough drop in a stock's price, capturing the exact severity of historical shocks like the COVID-19 crash or the 2022 election turbulence.

```python

# --- 1. PREPARATION: Ensure clean sorting and returns ---
df_master = df_master.sort_values(by=['Name', 'Date'])
df_master = df_master.dropna(subset=['Day Price'])

# Ensure Daily_Return exists
if 'Daily_Return' not in df_master.columns:
  df_master['Daily_Return'] = df_master.groupby('Name')['Day Price'].pct_change()


# --- 2. CALCULATE ANNUALIZED VOLATILITY ---
df_vol = (
    df_master.groupby(['Name', 'Sector'])['Daily_Return']
    .std()
    .reset_index(name='Daily_Volatility')
)
df_vol['Annualized_Volatility'] = df_vol['Daily_Volatility'] * np.sqrt(252)


# --- 3. CALCULATE COMPOUND ANNUAL GROWTH RATE (CAGR) ---
df_clean_price = df_master.dropna(subset=['Day Price'])
first_prices = df_clean_price.groupby('Name').first().reset_index()
last_prices = df_clean_price.groupby('Name').last().reset_index()

df_cagr = pd.merge(
    first_prices[['Name', 'Sector', 'Date', 'Day Price']],
    last_prices[['Name', 'Date', 'Day Price']],
    on='Name',
    suffixes=('_start', '_end'),
)

df_cagr['Years'] = (
    pd.to_datetime(df_cagr['Date_end'])
    - pd.to_datetime(df_cagr['Date_start'])
).dt.days / 365.25
df_cagr['Years'] = df_cagr['Years'].apply(lambda x: x if x > 0 else 1)

df_cagr['CAGR'] = (
    df_cagr['Day Price_end'] / df_cagr['Day Price_start']
) ** (1 / df_cagr['Years']) - 1


# --- 4. CALCULATE MAXIMUM DRAWDOWN (MaxDD) ---
def calculate_max_drawdown(group):
  # Calculate cumulative max to track peak prices
  prices = group['Day Price']
  peak = prices.cummax()
  drawdown = (prices - peak) / peak
  return drawdown.min()


df_maxdd = (
    df_master.groupby(['Name', 'Sector'])
    .apply(calculate_max_drawdown)
    .reset_index(name='Max_Drawdown')
)


# --- 5. MERGE ALL KPIS INTO A FINAL COMPREHENSIVE SUMMARY TABLE ---
kpi_summary = pd.merge(
    df_cagr[['Name', 'Sector', 'CAGR', 'Years']],
    df_vol[['Name', 'Annualized_Volatility']],
    on='Name',
)
kpi_summary = pd.merge(kpi_summary, df_maxdd[['Name', 'Max_Drawdown']], on='Name')

# Sort by CAGR to see top performers
kpi_summary = kpi_summary.sort_values(by='CAGR', ascending=False)

# Display the top 15 comprehensive company metrics
print('================================================================')
print('         NSE COMPANY-WISE COMPREHENSIVE KPI SUMMARY (2019-2025) ')
print('================================================================')
print(
    kpi_summary[
        [
            'Name',
            'Sector',
            'CAGR',
            'Annualized_Volatility',
            'Max_Drawdown',
            'Years',
        ]
    ]
    .head(15)
    .to_string(index=False)
)
```

**Sharpe Ratio (Risk-Adjusted Return)What it measures:**
- How much excess return you are getting for the extra volatility you endure.
- A higher Sharpe ratio means a better risk-adjusted investment.
- How to calculate:$$\text{Sharpe Ratio} = \frac{\text{CAGR} - \text{Risk-Free Rate}}{\text{Annualized Volatility}}$$

**Percentage of Zero-Volume Days (Liquidity Audit)**
- What it measures: Liquidity risk is a massive factor on frontier markets like the NSE. Some counters experience days (or weeks) with zero shares traded. This KPI tracks how "illiquid" a stock is.
- How to calculate: The percentage of trading days where Volume == 0 over the 7-year period.

**Average Daily Turnover (KES)**
- What it measures: Market depth and monetary liquidity by calculating how much actual capital flows through a stock daily.
- How to calculate: Daily Price multiplied by Volume (Day Price * Volume), then averaged per company.

```python
# --- 6. CALCULATE ADDITIONAL KPIS (Sharpe Ratio, Liquidity & Turnover) ---

# A. Risk-Free Rate assumption for Kenya (e.g., ~10% or 0.10)
risk_free_rate = 0.10
kpi_summary['Sharpe_Ratio'] = (
    kpi_summary['CAGR'] - risk_free_rate
) / kpi_summary['Annualized_Volatility']

# B. Calculate Zero-Volume Days Percentage (Liquidity Risk)
zero_vol_df = (
    df_master.groupby('Name')['Volume']
    .apply(lambda x: (x == 0).mean() * 100)
    .reset_index(name='Zero_Volume_Pct')
)

# C. Calculate Average Daily Turnover (Price * Volume)
df_master['Daily_Turnover'] = df_master['Day Price'] * df_master['Volume']
turnover_df = (
    df_master.groupby('Name')['Daily_Turnover']
    .mean()
    .reset_index(name='Avg_Daily_Turnover_KES')
)

# Merge these new KPIs into your master kpi_summary table
kpi_summary = pd.merge(kpi_summary, zero_vol_df, on='Name')
kpi_summary = pd.merge(kpi_summary, turnover_df, on='Name')

# View the updated comprehensive table with the new metrics
print('================================================================')
print('     ENHANCED NSE KPI SUMMARY (CAGR, Volatility, MaxDD, Sharpe)  ')
print('================================================================')
print(
    kpi_summary[
        [
            'Name',
            'Sector',
            'CAGR',
            'Annualized_Volatility',
            'Sharpe_Ratio',
            'Zero_Volume_Pct',
            'Avg_Daily_Turnover_KES',
        ]
    ]
    .head(15)
    .to_string(index=False)
)

# Optional: Export the final enriched table for your Tableau dashboards!
# kpi_summary.to_csv('nse_master_kpi_summary.csv', index=False)
```
